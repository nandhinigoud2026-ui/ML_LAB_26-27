```python
import numpy as np
x=np.array([3,2,1,4])
y=np.array([[5,6,7,8],[0,2,7,5]])
z=np.add(x,y)
print("addition of x and y:\n",z)
z=np.subtract(x,y)
print("subtraction of y from x:\n",z)
z=np.multiply(x,y)
print("element-wise multiplication of x and y:\n",z)
```
